```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

documents = [
    "Machine learning is a subfield of artificial intelligence.",
    "Natural language processing allows computers to understand text.",
    "Topic modeling finds hidden patterns in document collections.",
    "Latent Semantic Analysis is an early technique in topic modeling.",
    "Artificial intelligence and machine learning are closely related."
]

vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(documents)

lsa_model = TruncatedSVD(n_components=2, random_state=42)
lsa_topic_matrix = lsa_model.fit_transform(tfidf_matrix)

terms = vectorizer.get_feature_names_out()

for i, topic_vec in enumerate(lsa_model.components_):
    topic_terms = [terms[index] for index in topic_vec.argsort()[:-4:-1]]
    print(f"Topic {i}: {', '.join(topic_terms)}")
```
