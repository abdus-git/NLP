```python
import pandas as pd
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer

nltk.download('vader_lexicon', quiet=True)

data = {
    'review_id': [1, 2, 3, 4, 5],
    'review_text': [
        'The battery life is amazing and the screen is beautiful.',
        'Terrible customer service, and the product arrived broken.',
        'It works okay, but it is way too expensive for what it offers.',
        'Absolutely love this! Best purchase I have made this year.',
        'The hardware is decent, but the software is extremely buggy.'
    ]
}
df = pd.DataFrame(data)

sia = SentimentIntensityAnalyzer()

def get_sentiment_scores(text):
    scores = sia.polarity_scores(text)
    if scores['compound'] >= 0.05:
        return 'Positive'
    elif scores['compound'] <= -0.05:
        return 'Negative'
    else:
        return 'Neutral'

df['compound_score'] = df['review_text'].apply(lambda x: sia.polarity_scores(x)['compound'])
df['sentiment'] = df['review_text'].apply(get_sentiment_scores)

display(df)
```
