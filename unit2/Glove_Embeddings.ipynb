```python
import os
import numpy as np

EXTRACT_DIR = "glove_data"
os.makedirs(EXTRACT_DIR, exist_ok=True)
glove_file_path = os.path.join(EXTRACT_DIR, "glove.6B.50d.txt")

vocab_samples = {
    "science": np.random.uniform(-0.5, 0.5, 50),
    "technology": np.random.uniform(-0.5, 0.5, 50),
    "education": np.random.uniform(-0.5, 0.5, 50),
    "glove": np.random.uniform(-0.5, 0.5, 50),
    "embedding": np.random.uniform(-0.5, 0.5, 50)
}

with open(glove_file_path, "w", encoding="utf-8") as f:
    for word, vector in vocab_samples.items():
        vector_str = " ".join(map(str, vector))
        f.write(f"{word} {vector_str}\n")

print(f"Created a mock GloVe embedding file at: {glove_file_path}")
```

```python
def load_glove_embeddings(file_path):
    """Loads GloVe embeddings from a text file into a dictionary."""
    embeddings_index = {}
    print(f"Loading embeddings from {file_path}...")
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            values = line.split()
            word = values[0]
            coefs = np.asarray(values[1:], dtype='float32')
            embeddings_index[word] = coefs
    print(f"Loaded {len(embeddings_index)} word vectors.")
    return embeddings_index

glove_path = os.path.join(EXTRACT_DIR, "glove.6B.50d.txt")
embeddings = load_glove_embeddings(glove_path)
```

```python
def get_vector(word, embeddings_dict):
    """Retrieves the vector representation for a given word."""
    word_lower = word.lower()
    if word_lower in embeddings_dict:
        return embeddings_dict[word_lower]
    else:
        print(f"Word '{word}' not found in GloVe vocabulary.")
        return None

word = "science"
vector = get_vector(word, embeddings)

if vector is not None:
    print(f"\nVector representation for '{word}' (first 10 dimensions):\n", vector[:10])
    print(f"Vector shape: {vector.shape}")
```
