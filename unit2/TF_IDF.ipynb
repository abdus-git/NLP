```python
import numpy as np
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.pipeline import make_pipeline
from sklearn.metrics import classification_report, accuracy_score

categories = ['alt.atheism', 'soc.religion.christian', 'comp.graphics', 'sci.med']
train_data = fetch_20newsgroups(subset='train', categories=categories, shuffle=True, random_state=42)
test_data = fetch_20newsgroups(subset='test', categories=categories, shuffle=True, random_state=42)

nb_model = make_pipeline(TfidfVectorizer(), MultinomialNB())
nb_model.fit(train_data.data, train_data.target)
nb_predictions = nb_model.predict(test_data.data)

print("Naive Bayes Accuracy:", accuracy_score(test_data.target, nb_predictions))
print(classification_report(test_data.target, nb_predictions, target_names=test_data.target_names))

svm_model = make_pipeline(TfidfVectorizer(), LinearSVC(random_state=42))
svm_model.fit(train_data.data, train_data.target)
svm_predictions = svm_model.predict(test_data.data)

print("SVM Accuracy:", accuracy_score(test_data.target, svm_predictions))
print(classification_report(test_data.target, svm_predictions, target_names=test_data.target_names))
```
