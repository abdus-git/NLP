```python
# TF-IDF Implementation and Comparison with Bag of Words

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

documents = [
    "I love machine learning",
    "I love natural language processing",
    "Machine learning is interesting"
]


bow_vectorizer = CountVectorizer()

bow_matrix = bow_vectorizer.fit_transform(documents)

print("========== BAG OF WORDS ==========")

print("\nVocabulary:")
print(bow_vectorizer.vocabulary_)

print("\nFeature Names:")
print(bow_vectorizer.get_feature_names_out())

print("\nBoW Matrix:")
print(bow_matrix.toarray())



tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

print("\n\n========== TF-IDF ==========")

print("\nVocabulary:")
print(tfidf_vectorizer.vocabulary_)

print("\nFeature Names:")
print(tfidf_vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())



print("\n\n========== COMPARISON ==========")

print("\nBoW Representation:")
for i, vector in enumerate(bow_matrix.toarray()):
    print(f"Document {i + 1}: {vector}")

print("\nTF-IDF Representation:")
for i, vector in enumerate(tfidf_matrix.toarray()):
    print(f"Document {i + 1}: {vector}")


print("\n\n========== DIFFERENCE ==========")
print("BoW counts how many times each word occurs.")
print("TF-IDF gives higher importance to words that are")
print("important in a document but less common across documents.")
```
