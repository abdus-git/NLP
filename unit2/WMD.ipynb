```python
!pip install gensim pyemd

import gensim.downloader as api
from gensim.similarities import WmdSimilarity

model = api.load('word2vec-google-news-300')

doc1 = 'the quick brown fox jumps over the lazy dog'.lower().split()
doc2 = 'a fast brown fox leaps over a sleepy dog'.lower().split()

distance = model.wmdistance(doc1, doc2)
print(distance)
```
