```python
# Word2Vec Word Embeddings using Gensim
# Custom Corpus

!pip install gensim
from gensim.models import Word2Vec


corpus = [
    ["machine", "learning", "is", "interesting"],
    ["machine", "learning", "uses", "data"],
    ["deep", "learning", "is", "a", "part", "of", "machine", "learning"],
    ["natural", "language", "processing", "uses", "machine", "learning"],
    ["python", "is", "used", "for", "machine", "learning"],
    ["data", "science", "uses", "python", "and", "machine", "learning"],
    ["deep", "learning", "uses", "neural", "networks"],
    ["neural", "networks", "are", "used", "in", "deep", "learning"]
]


model = Word2Vec(
    sentences=corpus,
    vector_size=100,   # Size of word embedding
    window=5,          # Context window
    min_count=1,       # Include words appearing at least once
    workers=4,         # Number of CPU threads
    sg=1,              # 1 = Skip-gram, 0 = CBOW
    epochs=100
)


print("Vocabulary:")
print(model.wv.index_to_key)


word = "machine"

print(f"\nEmbedding for '{word}':")
print(model.wv[word])



print("\nWords similar to 'machine':")

similar_words = model.wv.most_similar("machine", topn=5)

for word, similarity in similar_words:
    print(f"{word}: {similarity:.4f}")

word1 = "machine"
word2 = "learning"

similarity = model.wv.similarity(word1, word2)

print(f"\nSimilarity between '{word1}' and '{word2}':")
print(f"{similarity:.4f}")


word = "python"

if word in model.wv:
    print(f"\n'{word}' exists in the vocabulary.")
else:
    print(f"\n'{word}' does not exist in the vocabulary.")
```
