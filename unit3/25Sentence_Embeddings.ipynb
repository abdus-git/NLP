```python
from transformers import AutoTokenizer, AutoModel
import torch

tokenizer_bert = AutoTokenizer.from_pretrained('bert-base-uncased')
model_bert = AutoModel.from_pretrained('bert-base-uncased')

sentences = ['Hello world', 'This is a sentence embedding example.']

inputs_bert = tokenizer_bert(sentences, padding=True, truncation=True, return_tensors='pt')
with torch.no_grad():
    outputs_bert = model_bert(**inputs_bert)

embeddings_bert = outputs_bert.last_hidden_state.mean(dim=1)
print(embeddings_bert.shape)
```

```python
from transformers import GPT2Tokenizer, GPT2Model
import torch

tokenizer_gpt2 = GPT2Tokenizer.from_pretrained('gpt2')
tokenizer_gpt2.pad_token = tokenizer_gpt2.eos_token
model_gpt2 = GPT2Model.from_pretrained('gpt2')

inputs_gpt2 = tokenizer_gpt2(sentences, padding=True, truncation=True, return_tensors='pt')
with torch.no_grad():
    outputs_gpt2 = model_gpt2(**inputs_gpt2)

embeddings_gpt2 = outputs_gpt2.last_hidden_state.mean(dim=1)
print(embeddings_gpt2.shape)
```
