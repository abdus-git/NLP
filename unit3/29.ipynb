```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

text = """The Hugging Face transformers library provides state-of-the-art machine learning architectures.
Among these, BART (Bidirectional and Auto-Regressive Transformers) is particularly effective for text summarization tasks.
It sequence-to-sequence model trained by denoising corrupted text.
Using the pipeline API, implementing this model requires only a few lines of code."""

prompt = f"Summarize the following text in one sentence:\n{text}\nSummary:"
output = generator(prompt, max_new_tokens=50, num_return_sequences=1)
print(output[0]['generated_text'])
```
