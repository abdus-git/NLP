```python
import nltk
from textblob import TextBlob
from nltk.sentiment.vader import SentimentIntensityAnalyzer

nltk.download('vader_lexicon', quiet=True)

text = "I absolutely love this product! It works perfectly and has changed my life. However, the shipping was a bit slow."

blob = TextBlob(text)
tb_polarity = blob.sentiment.polarity
tb_subjectivity = blob.sentiment.subjectivity

sia = SentimentIntensityAnalyzer()
vader_scores = sia.polarity_scores(text)

print(f"Text: {text}\n")
print(f"TextBlob Polarity: {tb_polarity}")
print(f"TextBlob Subjectivity: {tb_subjectivity}\n")
print(f"VADER Scores: {vader_scores}")
```
